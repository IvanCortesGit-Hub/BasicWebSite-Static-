```python
# Importar librerías necesarias
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score


# Cargar la base de datos desde el archivo Excel
df = pd.read_excel("estudiantes.xlsx")

# Mostrar las primeras filas
print("Primeras filas de la base de datos:")
print(df.head())

# Identificar dimensiones de la base
print("\nLa base de datos tiene:")
print(f"{df.shape[0]} observaciones (filas)")
print(f"{df.shape[1]} variables (columnas)")
```

```python
# Seleccionar la variable numérica (ejemplo: CGPA)
variable = "CGPA"

# Crear histograma
plt.figure(figsize=(8,5))
plt.hist(df[variable].dropna(), bins=20, edgecolor='black', color='skyblue')
plt.title(f"Histograma de {variable}")
plt.xlabel(variable)
plt.ylabel("Frecuencia")
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Crear boxplot
plt.figure(figsize=(6,5))
plt.boxplot(df[variable].dropna(), vert=True, patch_artist=True, boxprops=dict(facecolor="skyblue"))
plt.title(f"Boxplot de {variable}")
plt.ylabel(variable)
plt.grid(axis='y', alpha=0.7)
plt.show()
```

```python
# Crear variable dicotómica en función de CGPA
df["Rendimiento"] = df["CGPA"].apply(lambda x: 1 if x >= 7 else 0)

# Mostrar primeras filas con la nueva variable
print(df[["CGPA", "Rendimiento"]].head())

# Ver conteo de cada categoría
print("\nDistribución de la variable Rendimiento:")
print(df["Rendimiento"].value_counts())
```

```python
# Seleccionar variables explicativas (X) y variable objetivo (y)
X = df[["Academic Pressure", "Study Satisfaction", "Age"]].dropna()
y = df.loc[X.index, "Rendimiento"]

# Dividir en datos de entrenamiento y prueba (70% / 30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Entrenar el modelo de regresión logística
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Predicciones
y_pred = model.predict(X_test)

# Evaluación del modelo
print("Matriz de Confusión:\n", confusion_matrix(y_test, y_pred))
print("\nReporte de Clasificación:\n", classification_report(y_test, y_pred))
print("\nExactitud del modelo:", accuracy_score(y_test, y_pred))

```

```python
# Mostrar coeficientes de la regresión
coeficientes = pd.DataFrame({
    "Variable": X.columns,
    "Coeficiente": model.coef_[0]
})
print("\nCoeficientes del modelo:\n", coeficientes)
```
